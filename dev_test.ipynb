```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("raw_dataset/combined_dataset.csv")

# Select numerical columns for clustering
numerical_columns = [
    "r_cm", "v_cms", "temp", "peak", "snr", "width_us", "area",
    "power_percent", "snr_norm_r", "area_norm_r", "width_rel",
    "energy_density", "stability_std", "snr_variance", "echo_consistency",
    "gt_distance_cm"
]

missing = sorted(set(numerical_columns) - set(df.columns))
if missing:
    raise ValueError(f"Missing columns: {missing}")

# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[numerical_columns])

# KMeans clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(X_scaled)

print(df["cluster"].value_counts())
```
